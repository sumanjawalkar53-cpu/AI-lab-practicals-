## Step 2 - Make a map of rooms 🗺️

Our map has **5 rooms**, named **A, B, C, D, G**.
The robot starts at room **A** (Start) and wants to reach room **G** (Goal, the treasure 💎).

Rooms are joined by doors. We write down which rooms each room connects to.
Just run the box - it builds the map and draws it.


```python
# The map: each room lists the rooms you can walk to next.
map_rooms = {
    "A": ["B", "C"],   # from A you can go to B or C
    "B": ["D"],        # from B you can go to D
    "C": ["D"],        # from C you can go to D
    "D": ["G"],        # from D you can go to G
    "G": [],           # G is the treasure! No need to go further.
}

print("Our map:")
print("            A  (Start)")
print("           / \\")
print("          B   C")
print("           \\ /")
print("            D")
print("            |")
print("            G  (Treasure 💎)")
print()
print("Doors from each room:", map_rooms)
```

**Look at the map with your partner.** ✋
To get from **A** to **G**, the robot must pass through some middle rooms. Can you trace one path with your finger?


## Step 3 - Search Way 1: BFS (check nearest first) 🔵

**BFS** keeps a **to-do list** of rooms to visit. It always checks the room that has been
waiting **longest** (from the **front** of the list). This means it checks the **nearest**
rooms first - like ripples spreading out in water. 🌊

**Your turn:** find `# ✏️ YOUR TURN` and type **`0`** (the number zero).
`to_do.pop(0)` takes the room from the **front** of the list - that is what makes it BFS.


```python
def bfs(start, goal):
    to_do = [start]          # rooms waiting to be checked
    visited = []             # rooms we have already checked
    order = []               # the order we checked rooms (so we can see it)

    while to_do:
        # ✏️ YOUR TURN: take from the FRONT of the list. Type the number 0 inside pop( )
        room = to_do.pop(0)

        if room in visited:
            continue
        visited.append(room)
        order.append(room)

        if room == goal:
            return order      # found the treasure!

        for nxt in map_rooms[room]:
            to_do.append(nxt)
        print(to_do)
    return order

order = bfs("A", "G")
print("BFS checked the rooms in this order:", order)
print("Number of rooms BFS checked:", len(order))
```

If it printed an order ending in **G**, your BFS works! 🎉
Notice it checks **A, then B and C (the nearest), then D, then G** - nearest rooms first. 🔵


## Step 4 - Search Way 2: DFS (go deep first) 🟢

**DFS** uses the **same to-do list**, but takes the room from the **end** (the newest one).
This makes it dive **deep** down one path before trying others - like exploring one
tunnel all the way before coming back. 🕳️

**Your turn:** here you take from the **end** of the list. In Python, `to_do.pop()`
(with **nothing** inside the brackets) takes the **last** room. Type nothing between the
brackets - just delete the word `HERE`.


```python
def dfs(start, goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        # ✏️ YOUR TURN: take from the END of the list. Delete the word HERE so it reads: to_do.pop()
        room = to_do.pop()

        if room in visited:
            continue
        visited.append(room)
        order.append(room)

        if room == goal:
            return order

        for nxt in map_rooms[room]:
            to_do.append(nxt)
        print(to_do)
    return order

order = dfs("A", "G")
print("DFS checked the rooms in this order:", order)
print("Number of rooms DFS checked:", len(order))
```

See the difference? **DFS dives deep** down one path. **BFS spreads out** evenly.
Same map, same to-do list - the only change is **front vs end**! That tiny choice changes everything. 🤯


## Step 5 - Two ways to reach the treasure 🛣️

Look at the map again. There are **two** paths from A to G:

- **Path 1:**  A → B → D → G
- **Path 2:**  A → C → D → G

If every door is the same, both paths are equally good (same number of steps).
But what if some doors are **harder** to open and cost more? 🚪💪


```python
# How much each door costs to open (bigger number = harder door)
door_cost = {
    ("A", "B"): 1,
    ("B", "D"): 1,
    ("A", "C"): 5,   # this door is heavy and costs 5!
    ("C", "D"): 1,
    ("D", "G"): 1,
}

# A helper to add up the cost of a whole path
def path_cost(path):
    total = 0
    for i in range(len(path) - 1):
        door = (path[i], path[i + 1])
        total = total + door_cost[door]
    return total

path1 = ["A", "B", "D", "G"]
path2 = ["A", "C", "D", "G"]

print("Path 1  A -> B -> D -> G  costs:", path_cost(path1))
print("Path 2  A -> C -> D -> G  costs:", path_cost(path2))
```

See how Path 2 costs more, because the **A → C** door is heavy (cost 5)? 🚪💪
A smart robot should pick the **cheapest** path. Let's make it choose.


## Step 6 - Pick the CHEAPEST path 💰 (Uniform-Cost idea)

**Uniform-Cost Search** is a fancy name for a simple idea: **choose the path with the
lowest total cost.** Our robot compares the two paths and keeps the cheaper one.

**Your turn:** we want the **smaller** cost to win. Type **`<`** (the "less than" sign)
where it says `HERE`, so the robot picks the path that costs **less**.


```python
cost1 = path_cost(path1)
cost2 = path_cost(path2)

# ✏️ YOUR TURN: we want the cheaper path. Replace HERE with the < sign.
if cost1 < cost2:
    best_path = path1
    best_cost = cost1
else:
    best_path = path2
    best_cost = cost2

print("Path 1 cost:", cost1)
print("Path 2 cost:", cost2)
print()
print("The cheapest path is:", best_path, "with cost", best_cost, "💰")
```

Your robot picked the cheaper path! 🎉 That is the whole idea of Uniform-Cost Search:
when steps cost different amounts, **go the cheapest way**, not just the shortest-looking way.


## Step 7 - Compare the three searches 📊

Let's put it all together. Run the box to see a simple summary.


```python
bfs_order = bfs("A", "G")
dfs_order = dfs("A", "G")

print("RESULTS")
print("-" * 40)
print("BFS  (nearest first) checked:", len(bfs_order), "rooms ->", bfs_order)
print("DFS  (deep first)    checked:", len(dfs_order), "rooms ->", dfs_order)
print("Cheapest path:", best_path, "with cost", best_cost)
print("-" * 40)
print()
print("What we learned:")
print(" * BFS checks the nearest rooms first (spreads out).")
print(" * DFS dives deep down one path first.")
print(" * Cheapest path counts door costs and picks the lowest total.")
```

Great work! You ran three different searches and saw how each one behaves. 🎉


## Step 8 - Your turn to play! 🎮

In the box below, change the cost of the `("A", "C")` door to a **small** number like `1`,
